```python
# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/cars.csv')
df.columns = df.columns.str.strip() # Clean column names by stripping whitespace
df   #mpg in data is miles per galllon
```

Que 1 - Create a scatterplot for the following data:

*   mpg-hp
*   mpg-weightlbs
*   hp-mpg
*   hp-time-to-60

 with respect to brand column





```python
#mpg-hp
sns.scatterplot(x='mpg',y='hp', hue='brand', data=df)
# 'hue' uses the 'brand' column to color-code data points, differentiating brands visually.
# 'data' specifies the DataFrame (df) from which to draw the 'mpg', 'hp', and 'brand' columns.
plt.title('Hp vs MPG')
plt.xlabel('MPG')
plt.ylabel('HP')
plt.show()
```

```python
# mpg-weightlbs
sns.scatterplot(x='mpg',y='weightlbs', hue='brand', data=df)
# 'hue' uses the 'brand' column to color-code data points, differentiating brands visually.
# 'data' specifies the DataFrame (df) from which to draw the 'mpg', 'hp', and 'brand' columns.
plt.title('weightlbs vs MPG')
plt.xlabel('MPG')
plt.ylabel('weightlbs')
plt.show()
```

```python
 # hp-mpg
sns.scatterplot(x='hp',y='mpg', hue='brand', data=df)
# 'hue' uses the 'brand' column to color-code data points, differentiating brands visually.
# 'data' specifies the DataFrame (df) from which to draw the 'mpg', 'hp', and 'brand' columns.
plt.title('hp vs MPG')
plt.xlabel('hp')
plt.ylabel('mpg')
plt.show()
```

```python
# hp-time-to-60
sns.scatterplot(x='hp',y='time-to-60', hue='brand', data=df)
# 'hue' uses the 'brand' column to color-code data points, differentiating brands visually.
# 'data' specifies the DataFrame (df) from which to draw the 'mpg', 'hp', and 'brand' columns.
plt.title('hp vs time-to-60')
plt.xlabel('hp')
plt.ylabel('time-to-60')
plt.show()
```

```python
# subplots - a subplot means muliplt plot in one figure plane

fig=plt.subplots(2,2, figsize=(10,10))

# 1 plot
plt.subplot(2,2,1)
sns.scatterplot(x='mpg', y='hp', hue='brand', data=df)
plt.title('Hp vs MPG')
plt.xlabel('MPG')
plt.ylabel('HP')

# 2 plot
plt.subplot(2,2,2)
sns.scatterplot(x='mpg', y='weightlbs', hue='brand', data=df)
plt.title('weightlbs vs MPG')
plt.xlabel('MPG')
plt.ylabel('weightlbs')

# 3 plot
plt.subplot(2,2,3)
sns.scatterplot(x='hp',y='mpg', hue='brand', data=df)
plt.title('hp vs MPG')
plt.xlabel('hp')
plt.ylabel('mpg')

# 4 plot
plt.subplot(2,2,4)
sns.scatterplot(x='hp',y='time-to-60', hue='brand', data=df)
plt.title('hp vs time-to-60')
plt.xlabel('hp')
plt.ylabel('time-to-60')

plt.show()
```
