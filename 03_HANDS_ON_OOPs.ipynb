```python
# Que-1-abstarct class called Remotecontrol
# power_on and power_off abstract method
# creates a child class TV remote
# create an object of TV remote and demonstrate calling all methods

from abc import ABC , abstractmethod
class remotecontrol(ABC):
  @abstractmethod
  def power_on(self):
    pass

  @abstractmethod
  def power_off(self):
    pass

class TVremote(remotecontrol):
  def power_on(self):
    print("The TV is ON")

  def power_off(self):
    print("The Tv is off")

remote=TVremote()
remote.power_on()
remote.power_off()
```

```python
# Que-2-create a bankAccount:
# private variable:  __account_holder, __balance
# Methods :-
# a-deposit(amount)
# b-withdraw(amount) - only if enough balance
# c-get_balance

# create an object and perform deposit , withdraw and display balance, ensure balance is not accessed directly

class bankAccount:
  def __init__(self,holder,balance):
    self.__account_holder=holder
    self.__balance=balance

  def deposit(self,amount):
    if amount>0:
      self.__balance+=amount
      print(f"Successfully deposited {amount}")
    else:
       print("Invalid deposit amount")

  def withdraw(self, amount):
    if 0<amount<self.__balance:
      self.__balance-=amount
      print(f"successfullt withdrew : Rs{amount}")
    else :
      print("transaction failed: Insufficient balance or invalid amount.")

  def get_balance(self):    #As we cannot diretly access the private variate so we use this method to access this private variable __balance
    return self.__balance

acc=bankAccount("Rahul",1000)
print(acc.get_balance())

acc.deposit(500)
print(acc.get_balance())

acc.withdraw(500)
print(acc.get_balance())

```

```python
#Que-3-Create a class rectangle
#initialize with length and bredth
# implement methods:-
# a-area()-->returns the area
# b--perimeter()-> returns the perimeter

# instantiate the class and display both values

class rectangle:
  def __init__(self,length,breadth):
    self.length=length
    self.breadth=breadth

  def area(self):
    return self.length*self.breadth

  def perimeter(self):
    return 2*(self.length+self.breadth)

my_rect=rectangle(10,20)

my_rect.area()
print(f"the area of the rectangle is {my_rect.area()}")

my_rect.perimeter()
print(f"the perimeter od the rectangle is {my_rect.perimeter()}")
```

```python
#Que-5
#Library management system

#class Book:
  # attributes : title, availability status

#class library
  #Maintains a collection of books
  #a-add_books()->adds a book
  #b-issues a book()-> issues a book if available
  #c-return_book()->marks a books as returned

#simulates adding , issuing and returning books

class books:
  def __init__(self,title,availability):
    self.title=title
    self.availability=availability

class library:
  def __init__(self):
    self.books=[]

  def add_books(self,book):
    self.books.append(book)
    print(f"Added '{book.title}' to the library.")

  def issue_book(self,title):
    book_found = False
    for book in self.books:
      if book.title==title:
        book_found = True
        if book.availability:
          book.availability=False
          print(f"Successfully issued '{book.title}'.")
          return # Exit after issuing
        else:
          print(f"'{book.title}' is currently not available.")
          return # Exit, book found but not available
    if not book_found:
      print(f"'{title}' not found in the library.")

  def return_book(self, title):
    book_found = False
    for book in self.books:
      if book.title == title:
        book_found = True
        if not book.availability: # If it was issued (not available)
          book.availability = True
          print(f"Successfully returned '{book.title}'.")
          return # Exit after returning
        else:
          print(f"'{book.title}' was not issued, or is already available.")
          return # Exit, book found but already available
    if not book_found:
      print(f"'{title}' not found in the library.")

  def display_books(self):
    print("\nCurrent Library Status:")
    if not self.books:
        print("No books in the library.")
        return
    for book in self.books:
      status = "Available" if book.availability else "Issued"
      print(f"- {book.title}: {status}")

# Simulate adding, issuing and returning books
library_system = library()

book1 = books("The Great Gatsby", True)
book2 = books("1984", True)
book3 = books("To Kill a Mockingbird", True)

library_system.add_books(book1)
library_system.add_books(book2)
library_system.add_books(book3)

library_system.display_books()

print("\n--- Simulating issuing books ---")
library_system.issue_book("1984")
library_system.issue_book("The Great Gatsby")
library_system.issue_book("Non-existent Book")
library_system.issue_book("1984") # Try to issue again

library_system.display_books()

print("\n--- Simulating returning books ---")
library_system.return_book("1984")
library_system.return_book("The Great Gatsby")
library_system.return_book("Non-existent Book")
library_system.return_book("To Kill a Mockingbird") # Try to return a book that wasn't issued

library_system.display_books()
```

```python
#Que-11-Use abc module
# Create abstract class Animal with method sound()
# implement in Dog and Cat

from abc import ABC, abstractmethod

class Animal(ABC):
  @abstractmethod
  def sound(self):
    pass

class Dog(Animal):
  def sound(self):
    return "Bark"

class Cat(Animal):
  def sound(self):
    return "Meow"

animal=[Dog(),Cat()]
for i in animal:
  print(i.sound())
```

```python
#Que-13-Course enrollment system-
# student--> name
# Course-->course name, max capacity
#list of students
#mwthods--> enroll student(only if seats are available)
# display enrolled students

class student:
  def __init__(self,name):
    self.name=name

class course:
  def __init__(self,course_name,max_capacity):
    self.course_name=course_name
    self.max_capacity=max_capacity
    self.students=[]

  def enroll(self,student):
    if len(self.students)<self.max_capacity:
      self.students.append(student)

  def show_students(self):
    for s in self.students:
      print(s.name)

c=course("English Language",100)
s1=student("a")
s2=student("b")
s3=student("c")

c.enroll(s1)
c.enroll(s2)
c.enroll(s3)
c.show_students()
```

```python
#Que-14-Movie ticket booking:-
#Create a Movie Class
# attributes- name, total seats

#methods --> book_tickets(n)-> reduce ifseats are available
 # available_seats


class Movie:
  def __init__(self,name,seats):
    self.name=name
    self.seats=seats


  def book_tickets(self,n):
    if n<=self.seats:
      self.seats=self.seats-n
      print("Booking Done")
    else:
      print("Not enough seats")

  def available_seats(self):
   return self.seats

m=Movie("Dhurandhar",99)
m.book_tickets(20)
print(f'available seats are {m.available_seats()}')
```

```python
#Que-15-Create a class student
# attributes--.marks
#Methods--> a-average():-> calculate avg
#           b-result()--> if avg>40 pass else fAIL
#While creating the object for student class pass a list of marks

class student:
  def __init__(self,marks):
    self.marks=marks

  def average(self):
    return sum(self.marks)/len(self.marks)

  def result(self):
    return "pass" if self.average() >40 else "fail"

s=student([30,50,36,45])
print(s.average())
print(s.result())
```

```python
#Que-16- Bus reservation system:-
#create a bus class
#   total seats
#   booked seats

#Methods:
#  book_seat(n)
#  cancel_seat(n)
#  available seats()

class bus:
  def __init__(self,total):
    self.total=total
    self.booked=0


  def book_seat(self,n):
    if self.booked+n<=self.total:
      self.booked+=n
      print(f"{n} seats booked")
    else:
      print("Not enough seats")

  def cancel_seat(self,n):
    if n<=self.booked:
      self.booked=self.booked-n

  def available_seats(self):
    return self.total-self.booked

b=bus(60)
b.book_seat(30)
b.cancel_seat(13)
print(f'The available seats are {b.available_seats()}')
```

```python
#Que-17- Inventory System
# class Item--> name, qty
# methods --> add stock, reduce stock(if available)

class Item:
  def __init__(self,name,qty):
    self.name=name
    self.qty=qty

  def add_stock(self,n):
    self.qty+=n

  def reduce_stock(self,n):
    if n<=self.qty:
      self.qty-=n
    else:
      print("Not enough stock")

i=Item("soap",100)
i.add_stock(100)
i.reduce_stock(45)
print(i.qty)
```
