### Inheritance
Inheritance is a mechanism in object-oriented programming that allows a new class (subclass or child class) to inherit properties and behaviors from an existing class (superclass or parent class). This promotes code reusability and establishes a natural hierarchy between classes.

The following code demonstrates **Single Inheritance**.

- The `Car` class is the **parent (base) class**, defining common attributes (`brand`, `model`) and methods (`full_name`).
- The `ElectricCar` class is the **child (derived) class**, inheriting from `Car`.
- It uses `super().__init__(brand, model)` to call the constructor of the `Car` class, ensuring the parent's attributes are initialized.
- `ElectricCar` then adds its own specific attribute, `battery_size`.
- An instance of `ElectricCar` (`my_tesla`) can use both its own attributes/methods and those inherited from `Car` (e.g., `full_name()`).

```python
class Car:
  # This is the parent (base) class.
  # It defines common attributes and methods that will be inherited.
  def __init__(self,user_brand,user_model):
    self.brand=user_brand
    self.model=user_model

  def full_name(self):
    return f"{self.brand} {self.model}"

class ElectricCar(Car):
  # This is the child (derived) class, inheriting from the Car class.
  # It gains all public attributes and methods from Car.
  def __init__(self,brand, model, battery_size):
    # Call the constructor of the parent class (Car) using super().__init__.
    # This initializes the 'brand' and 'model' attributes from the Car class.
    super().__init__(brand,model)
    # Add a new attribute specific to ElectricCar.
    self.battery_size=battery_size

# Create an instance of ElectricCar.
# It requires arguments for both Car's constructor (brand, model) and its own (battery_size).
my_tesla=ElectricCar("Tesla","Model S","85KW")

# Access a method inherited from the Car class.
print(my_tesla.full_name())
# Access an attribute specific to the ElectricCar class.
print(my_tesla.battery_size)
```

```python
#"Private" Attributes (by convention): While Python doesn't have true private
# attributes like some other languages (you can still technically access the
# mangled name), the double underscore convention strongly indicates that these
# attributes are intended for internal use within the class and should not be
# accessed or modified directly from outside the class

class Car:
  def __init__(self,user_brand,user_model):
    # __brand uses name mangling (double underscore) to suggest it's 'private'.
    # It's a convention to indicate it should not be directly accessed from outside.
    self.__brand=user_brand
    self.model=user_model

  def get_brand(self):
    # This is a public method to access the 'private' __brand attribute.
    # It provides controlled access, which is a core concept of encapsulation.
    return self.__brand + "!"

  def full_name(self):
    return f"{self.__brand} {self.model}"

class ElectricCar(Car):
  def __init__(self,brand, model, battery_size):
    super().__init__(brand,model)
    self.battery_size=battery_size

my_tesla=ElectricCar("Tesla","Model S","85KW")
#print(my_tesla.brand) # Attempting to access __brand directly would cause an error (or need name mangling knowledge)
print(my_tesla.get_brand()) # Accessing the brand via the public getter method
```

```python
#Polymorphism
#-- demonstrate polymorphism by defining a method fuel_type in both car and electriccar classes, but with behaviours.

class Car:
  def __init__(self,user_brand,user_model):
    self.__brand=user_brand
    self.model=user_model

  def get_brand(self):
    return self.__brand + "!"

  def full_name(self):
    return f"{self.__brand} {self.model}"

  def fuel_type(self):
    # This is the implementation of fuel_type for a generic Car.
    return "Petrol or Diesel"

class ElectricCar(Car):
  def __init__(self,brand, model, battery_size):
    super().__init__(brand,model)
    self.battery_size=battery_size

  def fuel_type(self):
    # This method OVERRIDES the fuel_type method from the parent Car class.
    # It provides a specific implementation for ElectricCar, demonstrating polymorphism.
    return "Electric Charge"

my_tesla=ElectricCar("Tesla","Model S","85KW")
# When fuel_type is called on my_tesla (an ElectricCar object),
# Python calls the ElectricCar's version of fuel_type.
print(my_tesla.fuel_type())

Safari=Car("Tata","Safari")
# When fuel_type is called on Safari (a Car object),
# Python calls the Car's version of fuel_type.
print(Safari.fuel_type())
```

```python
# CLASS INHERITANCE AND ISINSTANCE() FUNCTION
# Demonstrate the use of isinstance() to check if my_tesla is an instance of Car and ElectricCar

# A class defines common attributes and methods.
class Car:
  total_cars=0

  def __init__(self,user_brand,user_model):
    self.__brand=user_brand
    self.__model=user_model
    Car.total_cars+=1

  def get_brand(self):
    return self.__brand + "!"

  def full_name(self):
    return f"{self.__brand} {self.__model}"

  def fuel_type(self):
    return "Petrol or Diesel"

  @staticmethod
  def general_description():
    return "Cars are vehicles that transport people and goods."

  @property
  def model(self):
    return self.__model


# ElectricCar inherits from Car, meaning it gains all public attributes and methods from Car.
class ElectricCar(Car):
  def __init__(self,brand, model, battery_size):
    # super().__init__() calls the constructor of the parent class (Car)
    # to initialize inherited attributes like brand and model.
    super().__init__(brand,model)
    self.battery_size=battery_size

  def fuel_type(self):
    return "Electric Charge"


# Create an instance of ElectricCar
my_tesla=ElectricCar("Tesla","Model S","85KW")

# isinstance(object, classinfo) checks if an object is an instance of a specified class
# or a class derived from it. It returns True or False.

# Check if my_tesla is an instance of Car (its parent class)
print(f"Is my_tesla an instance of Car? {isinstance(my_tesla, Car)}")

# Check if my_tesla is an instance of ElectricCar (its own class)
# Fixed typo: 'Electriccar' to 'ElectricCar'
print(f"Is my_tesla an instance of ElectricCar? {isinstance(my_tesla, ElectricCar)}")
```

```python
# Multiple Inheritance
# --Create two classes Battery and Engine, and let the ElectricCar Class inherit
# from both, demonstrating multiple inheritance

# Base class Car, defining general car properties
class Car:
  total_cars=0

  def __init__(self,user_brand,user_model):
    self.__brand=user_brand
    self.__model=user_model
    Car.total_cars+=1

  def get_brand(self):
    return self.__brand + "!"

  def full_name(self):
    return f"{self.__brand} {self.__model}"

  def fuel_type(self):
    return "Petrol or Diesel"

  @staticmethod
  def general_description():
    return "Cars are vehicles that transport people and goods."

  @property
  def model(self):
    return self.__model


# ElectricCar inherits from Car (single inheritance example, not directly used in the multiple inheritance below)
class ElectricCar(Car):
  def __init__(self,brand, model, battery_size):
    super().__init__(brand,model)
    self.battery_size=battery_size

  def fuel_type(self):
    return "Electric Charge"


# A component class for battery information
class Battery:
  def battery_info(self):
    return "This car uses an electric battery."

# A component class for engine information
class Engine:
  def engine_info(self):
    # Fixed: Added the missing closing double quote for the string literal.
    return "This car has an electric motor (engine)."

# ElectricCarTwo demonstrates multiple inheritance.
# It inherits from Battery, Engine, and Car.
# The order of inheritance matters for method resolution order (MRO).
class ElectricCarTwo(Battery, Engine, Car):
  # When using multiple inheritance and parent classes have __init__ methods
  # that require arguments, you need to explicitly call them.
  # Python's `super()` works with MRO, but for complex scenarios with args,
  # direct calls to parent __init__ are often clearer.
  def __init__(self, brand, model, battery_size):
    # Initialize attributes from the Car class
    Car.__init__(self, brand, model)
    # Battery and Engine do not have __init__ methods that take arguments in this example,
    # so they don't need explicit calls here unless they had their own __init__.
    # If they had __init__ methods, you would call them like: Battery.__init__(self), Engine.__init__(self)
    self.battery_size = battery_size


# Create an instance of ElectricCarTwo
# This object now has properties and methods from Car, Battery, and Engine.
my_new_tesla = ElectricCarTwo("Tesla","Model S", "100KWH")

# Access methods inherited from Battery and Engine
print(f"Battery Info: {my_new_tesla.battery_info()}")
print(f"Engine Info: {my_new_tesla.engine_info()}")

# Access methods inherited from Car
print(f"Full Name: {my_new_tesla.full_name()}")
print(f"Car Model: {my_new_tesla.model}")
```
