Que 3 - Create a nested list and a dictionary(store thevalues in the form of a List. --Convert atleast 3 keys. -- Convert these into pandas DataFrame

```python
import pandas as pd

# Create a nested list
nested_list = [
    ['Apple', 10, 1.50],
    ['Banana', 20, 0.75],
    ['Cherry', 15, 2.00],
    ['Date', 25, 1.20]
]

# Convert nested list to DataFrame
df_from_list = pd.DataFrame(nested_list, columns=['Fruit', 'Quantity', 'Price'])
print("DataFrame from nested list:")
display(df_from_list)
```

```python
# Create a dictionary with list values (at least 3 keys)
my_dict = {
    'City': ['New York', 'Los Angeles', 'Chicago'],
    'Population': [8419000, 3980000, 2705000],
    'State': ['NY', 'CA', 'IL'],
    'Area_sq_mi': [302.6, 468.7, 234.0]
}

# Convert dictionary to DataFrame
df_from_dict = pd.DataFrame(my_dict)
print("\nDataFrame from dictionary:")
display(df_from_dict)
```

```python
# Rename the column name from "Car-model"  to " Model"
```

```python
df_from_dict.rename(columns = {"City":"Model"}, inplace=True)
```

```python
df_from_dict
```

```python
# Select all the records  where population is greater than 3000000
# Select all the records where the model is Chicago and population is greater than 2500000
# Select the records where the model is New York or population is greater than 3900000
```

```python
df_from_dict.loc[df_from_dict["Population"]>3000000]
```

```python
df_from_dict.loc[(df_from_dict["Model"] == "Chicago") & (df_from_dict["Population"] > 2500000)]
```

```python
# Select the records where the model is New York or population is greater than 3900000
df_from_dict.loc[(df_from_dict["Model"] == "New York") | (df_from_dict["Population"] > 3900000)]
```
