HOW avgMonthlyHours is affecting the employee attrition rate ?


```python
sns.histplot(data=df,x='avgMonthlyHours',hue='left')
plt.show()                # 0 - employee still working in th ecompany
                          # 1 - employee who left the country
```

**Inference:**

Q- How the Promotion in the last 5 years is affecting the employee attrition rate.

**Inference:**

**How the Salary is affecting the employee attrition Rate.**
