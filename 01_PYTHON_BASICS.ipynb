A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself. In simpler terms, it's a number that can only be divided evenly by 1 and itself.

```python
import math
print('Prime numbers between 2 and 20 are:')

for num in range(2, 21):
    # Prime numbers are greater than 1
    # Iterate from 2 to num - 1
    for i in range(2,int(math.sqrt(num))+1 ):
        # If num is divisible by any number between 2 and num-1, it is not prime
        if (num % i) == 0:
            break
    else:
        print(num)
```

#### Explanation of the Prime Number Code:

1.  **`print('Prime numbers between 2 and 20 are:')`**
    *   This line simply prints a descriptive header for the output.

2.  **`for num in range(2, 21):`**
    *   This is the **outer loop**. It iterates through each number from 2 up to (but not including) 21. Each `num` in this range will be checked to see if it's a prime number.

3.  **`for i in range(2, num):`**
    *   This is the **inner loop**. For each `num` from the outer loop, this loop iterates through numbers starting from 2 up to `num - 1`. These `i` values are potential divisors of `num`.

4.  **`if (num % i) == 0:`**
    *   Inside the inner loop, this condition checks if `num` is perfectly divisible by `i` (i.e., the remainder of the division is 0).
    *   If `num` is divisible by any `i` (other than 1 and itself), it means `num` has a divisor other than 1 and itself, so it cannot be a prime number.

5.  **`break`**
    *   If `num % i == 0` is true, it means we found a divisor for `num`. We immediately know `num` is not prime, so there's no need to check further divisors for this `num`. The `break` statement exits the **inner loop**.

6.  **`else:`**
    *   This `else` block is associated with the **inner `for` loop**, not the `if` statement. An `else` block for a `for` loop is executed *only if the loop completes without encountering a `break` statement*.
    *   If the inner loop completes without a `break`, it means `num` was not divisible by any `i` in the range `(2, num)`. This implies that `num` has no divisors other than 1 and itself (since we started checking from 2 and went up to `num-1`).

7.  **`print(num)`**
    *   If the `else` block is executed, it confirms `num` is a prime number, and thus `num` is printed to the console.

To check for an Armstrong number, you need to know the total number of digits first so you can raise each digit to that power. This is why it often requires either two loops (one to count digits, one to sum powers) or converting the number to a string to easily get its length.

```python
# Armstrong number -- a number that is equal to the sum of digits raised to the power of the total number of digits
# example - 153 --> 1**3 + 5**3 + 3**3 = 1 + 125 + 27 = 153

original_num = int(input("Enter the number to check if it's an Armstrong number: "))

# Store the number in a temporary variable to use it later
temp_num = original_num
count = 0

# First pass: Count the number of digits
while temp_num > 0:
    temp_num //= 10  # Integer division to remove the last digit
    count += 1

print(f"Length of number is: {count}")

# Reset temp_num to the original number for the second pass
temp_num = original_num
sum_of_powers = 0

# Second pass: Calculate the sum of digits raised to the power of 'count'
while temp_num > 0:
    digit = temp_num % 10  # Get the last digit
    sum_of_powers += digit ** count
    temp_num //= 10

# Compare the sum with the original number
if original_num == sum_of_powers:
    print(f"{original_num} is an Armstrong number.")
else:
    print(f"{original_num} is not an Armstrong number.")
```

```python
#uestion - write a pyhton program to count the frequency of each number in the list

n=int(input("Enter the number of elements:"))
num=[]
for i in range(n):
  x=int(input(f"Enter element {i+1}: "))
  num.append(x)

# Using a dictionary to store frequencies
frequency = {}
for item in num:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1

print("Frequency of each number:")
for key, value in frequency.items():
    print(f"{key}: {value}")
```

2. Create an empty list.

3. Take each number and put it into the list.

4. Create an empty dictionary called frequency.

5. Go through the list:
      - If the number already exists → increase its count.
      - If it doesn't exist → add it with count 1.

```python
# Que - Count how many times each character appears in a string

str1=(input("Enter the string :"))

# Using a dictionary to store frequencies
frequency = {}
for ch in str1:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1

print("Frequency of each number:")
for key, value in frequency.items():
    print(f"{key} --> {value}")
```

```python
# Calculate the averag marks of all the students
# find the student who scored the highest marks and print the name of the student

marks={
    "Yash":89,
    "Kanishka":78,
    "Srikant":77,
    "Hariom":98,
    "Nirvan":99
}

total=0
top_student=""
highest=0

for name,score in marks.items():
  total+=score

  if score>highest:
    highest=score
    top_student=name

average=total/len(marks)
print("average marks : ",average)
print("top student : ",top_student)

```

```python
n=int(input("Enter the number of elements:"))
num=[]
for i in range(n):
  x=(input(f"Enter element {i+1}: "))
  num.append(x)
longest=num[0]

for word in num:
  if len(word)>len(longest):
    longest=word

print("The longest string is :",longest)
```

```python
# Sort the list on the basis of the length of the string

n=int(input("Enter the number of elements:"))
num=[]
for i in range(n):
  x=input(f"Enter element {i+1}: ")
  num.append(x)

res=sorted(num,key=lambda x:len(x))
print(res)
```

```python
n = int(input("Enter the number of elements: "))
input_list = []
for i in range(n):
  item = input(f"Enter element {i+1}: ")
  input_list.append(item)

# Using a dictionary to count element frequencies
counts = {}
for item in input_list:
    counts[item] = counts.get(item, 0) + 1

duplicates = []
for item, count in counts.items():
    if count > 1:
        duplicates.append(item)

if duplicates:
    print("Duplicate elements in the list are:", duplicates)
else:
    print("No duplicate elements found in the list.")
```
